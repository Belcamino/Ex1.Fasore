```python
import sys

def tartaglia(n):
    """
    Genera e stampa il Triangolo di Tartaglia (noto anche come Triangolo di Pascal)
    fino alla riga 'n'. Mostra anche i coefficienti binomiali C(N, k).
    """
    # Inizializziamo la prima riga del triangolo
    riga = [1]

    for N in range(n + 1):
        # Trasformiamo i numeri in stringhe e li uniamo con uno spazio
        riga_testo = " ".join(map(str, riga))

        # Centriamo il testo per dare la classica forma a piramide (triangolo)
        testo_centrato = riga_testo.center(4 * n)

        # Stampiamo la riga corrente con l'indicazione dei coefficienti binomiali C(N, k)
        print(f"{testo_centrato} <- C({N},k), k = 0..{N}")

        # Calcoliamo la riga successiva sommando gli elementi adiacenti della riga corrente
        # zip(riga, riga[1:]) accoppia gli elementi adiacenti: (riga[0], riga[1]), (riga[1], riga[2]), ecc.
        riga = [1] + [a + b for a, b in zip(riga, riga[1:])] + [1]

def main():
    # Definiamo il numero di righe da generare
    numero_righe = 7
    tartaglia(numero_righe)

# Il punto di ingresso standard per l'esecuzione dello script
if __name__ == "__main__":
    main()
```
